```python
result = run_rar_workflow(
    training_pool_path=training_pool_path,
    official_test_path=official_test_path,
    manifest_path=manifest_path,
    language=language,
    rar_config_path=Path("../config/rar.yaml"),
    models_config_path=Path("../config/models.yaml"),
    output_dir=output_dir,
)
result
```
