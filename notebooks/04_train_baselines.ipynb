# Train a supervised baseline

Train BERT or PURE using the manifest-defined train/development split and evaluate the untouched official test set.

```python
from pathlib import Path

from clinical_re_bilingual.schema import Language
from clinical_re_bilingual.workflows import run_baseline_workflow
```

```python
language = Language.ENGLISH
baseline = "pure"  # Use "bert" for the sequence-classification baseline.
code = language.value
count = run_baseline_workflow(
    baseline=baseline,
    training_pool_path=Path(f"../data/processed/{code}_training_pool.tsv"),
    official_test_path=Path(f"../data/processed/{code}_official_test.tsv"),
    manifest_path=Path(f"../data/manifests/{code}_split.json"),
    language=language,
    baselines_config_path=Path("../config/baselines.yaml"),
    models_config_path=Path("../config/models.yaml"),
    output_dir=Path(f"../checkpoints/baselines/{baseline}/{code}"),
)
count
```
