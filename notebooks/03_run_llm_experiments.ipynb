# Run an LLM experiment

Run one reproducible provider/model/language/retriever/shot/strategy configuration. The output contains IDs and labels only.

```python
from pathlib import Path

from clinical_re_bilingual.prompting import PromptStrategy
from clinical_re_bilingual.workflows import run_inference_workflow
```

```python
language = "en"
provider = "gemini"
model_name = "gemini-1.5-flash-latest"
num_shots = 5
strategy = PromptStrategy.STANDARD
retriever = "rar"
```

```python
count = run_inference_workflow(
    training_pool_path=Path(f"../data/processed/{language}_training_pool.tsv"),
    official_test_path=Path(f"../data/processed/{language}_official_test.tsv"),
    rankings_path=Path(f"../checkpoints/{retriever}/{language}/rankings.jsonl"),
    provider=provider,
    model_name=model_name,
    num_shots=num_shots,
    strategy=strategy,
    temperature=0.0,
    output_path=Path(
        f"../results/metrics/{language}_{model_name}_{retriever}_{num_shots}_{strategy.value}.jsonl"
    ),
)
count
```
